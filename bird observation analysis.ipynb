```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import os

print("Current Working Directory:")
print(os.getcwd())

print("\nFiles:")
print(os.listdir())
```

```python
import pandas as pd

forest_df = pd.read_csv("Bird_Monitoring_Data_Forest.csv")
grassland_df = pd.read_csv("Bird_Monitoring_Data_GRASSLAND.csv")

print("Forest data shape:", forest_df.shape)
print("Grassland data shape:", grassland_df.shape)
```

```python
forest_df.head()
```

```python
grassland_df.head()
```

```python
forest_df.info()
```

```python
grassland_df.info()
```

```python
print(forest_df.columns.tolist())
```

```python
print(grassland_df.columns.tolist())
```

```python
forest_df.columns = forest_df.columns.str.strip()
grassland_df.columns = grassland_df.columns.str.strip()
```

```python
forest_df = forest_df.loc[:, ~forest_df.columns.str.contains("^Unnamed")]
grassland_df = grassland_df.loc[:, ~grassland_df.columns.str.contains("^Unnamed")]
```

```python
print(forest_df.columns.tolist())
```

```python
forest_df["Habitat"] = "Forest"
grassland_df["Habitat"] = "Grassland"
```

```python
forest_df["Habitat"].value_counts()
```

```python
grassland_df["Habitat"].value_counts()
```

```python
combined_df = pd.concat(
    [forest_df, grassland_df],
    ignore_index=True
)
```

```python
print("Combined Shape:", combined_df.shape)
```

```python
combined_df.head()
```

```python
combined_df["Habitat"].value_counts()
```

```python
missing = combined_df.isnull().sum()

print(missing[missing > 0])
```

```python
missing_percentage = (
    combined_df.isnull().sum() /
    len(combined_df) * 100
)

print(missing_percentage[missing_percentage > 0])
```

```python
print("Duplicate rows:", combined_df.duplicated().sum())
```

```python
combined_df = combined_df.drop_duplicates()
```

```python
print("Shape after removing duplicates:", combined_df.shape)
```

```python
combined_df["Date"] = pd.to_datetime(
    combined_df["Date"],
    errors="coerce"
)
```

```python
combined_df["Date"].dtype
```

```python
combined_df["Month"] = combined_df["Date"].dt.month

combined_df["Month_Name"] = combined_df["Date"].dt.month_name()

combined_df["Season"] = combined_df["Date"].dt.month.map({
    12: "Winter",
    1: "Winter",
    2: "Winter",
    3: "Spring",
    4: "Spring",
    5: "Spring",
    6: "Summer",
    7: "Summer",
    8: "Summer",
    9: "Autumn",
    10: "Autumn",
    11: "Autumn"
})
```

```python
total_observations = len(combined_df)

print("Total Observations:", total_observations)
```

```python
total_species = combined_df["Scientific_Name"].nunique()

print("Total Species:", total_species)
```

```python
print(combined_df["Habitat"].nunique())
```

```python
print(combined_df["Location_Type"].nunique())
```

```python
habitat_counts = combined_df["Habitat"].value_counts()

print(habitat_counts)
```

```python
plt.figure(figsize=(8,5))

habitat_counts.plot(kind="bar")

plt.title("Bird Observations by Habitat")
plt.xlabel("Habitat")
plt.ylabel("Number of Observations")
plt.xticks(rotation=0)

plt.show()
```

```python
species_counts = (
    combined_df["Common_Name"]
    .value_counts()
    .head(10)
)

print(species_counts)
```

```python
plt.figure(figsize=(10,6))

species_counts.sort_values().plot(kind="barh")

plt.title("Top 10 Most Observed Bird Species")
plt.xlabel("Number of Observations")
plt.ylabel("Bird Species")

plt.show()
```

```python
species_by_habitat = (
    combined_df.groupby("Habitat")["Scientific_Name"]
    .nunique()
)

print(species_by_habitat)
```

```python
plt.figure(figsize=(8,5))

species_by_habitat.plot(kind="bar")

plt.title("Unique Bird Species by Habitat")
plt.xlabel("Habitat")
plt.ylabel("Number of Unique Species")
plt.xticks(rotation=0)

plt.show()
```

```python
year_counts = combined_df["Year"].value_counts().sort_index()

print(year_counts)
```

```python
plt.figure(figsize=(10,5))

year_counts.plot(kind="line", marker="o")

plt.title("Bird Observations by Year")
plt.xlabel("Year")
plt.ylabel("Observations")

plt.show()
```

```python
season_counts = combined_df["Season"].value_counts()

print(season_counts)
```

```python
plt.figure(figsize=(8,5))

season_counts.plot(kind="bar")

plt.title("Bird Observations by Season")
plt.xlabel("Season")
plt.ylabel("Observations")

plt.xticks(rotation=0)

plt.show()
```

```python
combined_df["Temperature"].describe()
```

```python
temperature_activity = (
    combined_df.groupby("Temperature")
    .size()
    .reset_index(name="Observations")
)
```

```python
fig = px.scatter(
    temperature_activity,
    x="Temperature",
    y="Observations",
    title="Bird Observations vs Temperature"
)

fig.show()
```

```python
humidity_activity = (
    combined_df.groupby("Humidity")
    .size()
    .reset_index(name="Observations")
)

fig = px.scatter(
    humidity_activity,
    x="Humidity",
    y="Observations",
    title="Bird Observations vs Humidity"
)

fig.show()
```

```python
id_method_counts = combined_df["ID_Method"].value_counts()

print(id_method_counts)
```

```python
plt.figure(figsize=(10,5))

id_method_counts.plot(kind="bar")

plt.title("Bird Identification Methods")
plt.xlabel("Identification Method")
plt.ylabel("Observations")

plt.xticks(rotation=45)

plt.show()
```
